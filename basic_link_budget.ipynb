```python
def FSPL_db(d, f):
    """
    Calculate Free Space Path Loss (FSPL) in dB.

    Parameters:
    d (float): Distance between transmitter and receiver in meters.
    f (float): Frequency of the signal in Hz.

    Returns:
    float: FSPL in dB.
    """
    c = 3e8  # Speed of light in m/s
    fspl = 20 * np.log10(d) + 20 * np.log10(f) - 147.55
    return fspl
```

```python
f_carrier = 433e6
mean_distance = 600e3
tx_power_dbm = 20  # Transmitted power in dBm
tx_antenna_gain_dbi = 0 
rx_antenna_gain_dbi = 3
polarization_loss_db = 3
additional_losses_db = 2
```

```python

power_budget = [
  ('Transmitted Power', tx_power_dbm),
  ('Transmitter Antenna Gain', tx_antenna_gain_dbi),
  ('Free Space Path Loss', -FSPL_db(d = mean_distance, f = f_carrier)),
  ('Receiver Antenna Gain', rx_antenna_gain_dbi),
  ('Polarization Loss', -polarization_loss_db),
  ('Additional Losses', -additional_losses_db),
]
```

```python
power_received_dbm = sum(value for _, value in power_budget)
print(f"Received Power: {power_received_dbm:.2f} dBm\n")
```

```python
noise_figure_db = 2.5
T0 = 290

noise_power_dbm = 10 * np.log10(1.38e-23 * T0 * 1e6) + noise_figure_db
print(f"Noise Power: {noise_power_dbm:.2f} dBm\n")
print(f"Signal-to-Noise Ratio (SNR): {power_received_dbm - noise_power_dbm:.2f} dB\n")
```
